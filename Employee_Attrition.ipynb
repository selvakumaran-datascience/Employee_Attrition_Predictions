```python
# 4) Encoding dataframe
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
employee_data['Attrition'] = le.fit_transform(employee_data['Attrition'])
```

```python
sns.countplot(x="Attrition",data=employee_data,palette="coolwarm")
plt.title('Attrition Distribution')
plt.show()
```

```python
order_col = ['Gender','MaritalStatus','OverTime','BusinessTravel','JobRole','Department']
```

```python
from sklearn.preprocessing import OrdinalEncoder
order = {
    'Gender':['Female', 'Male'],
    'MaritalStatus':['Single', 'Married', 'Divorced'],
    'OverTime':['Yes', 'No'],
    'JobRole': ['Human Resources', 'Laboratory Technician', 'Healthcare Representative', 'Sales Representative', 'Sales Executive', 'Manager', 'Manufacturing Director', 'Research Scientist', 'Research Director'],
    'BusinessTravel': ['Non-Travel', 'Travel_Rarely', 'Travel_Frequently'],
    'Department': ['Sales', 'Research & Development', 'Human Resources']
}

order_col = list(order.keys())

ordinal_encoder = OrdinalEncoder(categories= [order[col] for col in order_col])
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(transformers=[('ordinal',ordinal_encoder,order_col)],remainder='passthrough')
```

```python
x = employee_data.drop(columns=['Attrition'])
y = employee_data['Attrition']
```

```python
df_e = preprocessor.fit_transform(x)
```

```python
employee_data = pd.DataFrame(df_e,columns=preprocessor.get_feature_names_out())
```

```python
employee_data.columns = (
    employee_data.columns
    .str.replace("ordinal","")
    .str.replace("nominal","")
    .str.replace("remainder","") 
    .str.replace("__","")               
)
```

```python
employee_data.head()
```

```python
employee_data_clean=pd.concat([y,employee_data],axis=1)
```

```python
employee_data_clean.head()
```

```python
employee_data_clean.info()
```

```python
plt.figure(figsize=(25,15))
sns.heatmap(employee_data_clean.corr(),annot=True,vmin=-1,vmax=1)
plt.show()
```

```python
num_cols = employee_data_clean.select_dtypes(include='number').columns
num_cols
```

```python
# 6) Finding Outliers
outlier_summary = {}

for col in num_cols:
  Q1 = employee_data_clean[col].quantile(0.25)
  Q3 = employee_data_clean[col].quantile(0.75)
  IQR = Q3-Q1
  lower_limit = Q1-1.5*IQR
  upper_limit = Q3+1.5*IQR

  outlier = employee_data_clean[(employee_data_clean[col] < lower_limit) | (employee_data_clean[col] > upper_limit)]
  print(f"{col} has {len(outlier)} outliers")
  outlier_percentage = round(len(outlier)/len(employee_data_clean)*100,2)

  outlier_summary[col] = {
      'Q1':Q1,
      'Q3':Q3,
      'IQR':IQR,
      'lower_limit':lower_limit,
      'upper_limit':upper_limit,
      'outliers count':len(outlier),
      'outlier_percentage %':outlier_percentage
  }
```

```python
outlier_df = pd.DataFrame(outlier_summary).T

outlier_df
```

```python
# 8) Select Features And Target
x = employee_data_clean.drop(columns=['Attrition'])
y = employee_data_clean['Attrition']

print("Shape of X:", x.shape)
print("Shape of y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
# 9) Train Test Split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
print("Shape of X_train:", x_train.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of X_test:", x_test.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
models = {
    'Logistic Regression': (
        LogisticRegression(),
        {
            'model__C': [0.01, 0.1, 1, 10, 100],
            'model__penalty': ['l1', 'l2'],
            'model__solver': ['liblinear']
        }
    ),

    'KNN': (
        KNeighborsClassifier(),
        {
            'model__n_neighbors': [3, 5, 7, 9, 11, 15],
            'model__weights': ['uniform', 'distance'],
            'model__metric': ['euclidean', 'manhattan']
        }
    ),

    'Decision Tree': (
        DecisionTreeClassifier(random_state=42),
        {
            'model__max_depth': [3, 5, 7, 10, 15, None],
            'model__min_samples_split': [2, 5, 10, 20],
            'model__min_samples_leaf': [1, 2, 4, 8],
            'model__criterion': ['gini', 'entropy']
        }
    ),

    'Random Forest': (
        RandomForestClassifier(random_state=42),
        {
            'model__n_estimators': [100, 200, 300, 500],
            'model__max_depth': [5, 10, 20, None],
            'model__min_samples_split': [2, 5, 10],
            'model__min_samples_leaf': [1, 2, 4],
            'model__max_features': ['sqrt', 'log2'],
            'model__criterion': ['gini', 'entropy']
        }
    )
}
```

```python
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import *
```

```python
results = {}

for name, (model, params) in models.items():

    # Pipeline
    pipe = ImbPipeline([
        ('scaler', StandardScaler()),
        ('smote', SMOTE(random_state=42)),
        ('model', model)
    ])

    # Grid Search
    grid = GridSearchCV(
        estimator=pipe,
        param_grid=params,
        cv=5,
        scoring='f1',
        n_jobs=-1,
        verbose=1
    )

    # Train
    grid.fit(x_train, y_train)

    # Best Model
    best_model = grid.best_estimator_

    # Prediction
    y_pred = best_model.predict(x_test)

    # Metrics
    results[name] = {
        "Best Parameters": grid.best_params_,
        "CV Score": round(grid.best_score_, 4),
        "Accuracy": round(accuracy_score(y_test, y_pred), 4),
        "Precision": round(precision_score(y_test, y_pred), 4),
        "Recall": round(recall_score(y_test, y_pred), 4),
        "F1 Score": round(f1_score(y_test, y_pred), 4)
    }
```

```python
result_df = pd.DataFrame(results).T
result_df
```

```python
# 10) Best Model
print(best_model)
```

```python
print(classification_report(y_test,y_pred))
```

```python
y_prob =best_model.predict_proba(x_test)
fpr,tpr,thresholds = roc_curve(y_test,y_prob[:,1])
auc_score = roc_auc_score(y_test,y_prob[:,1])
plt.plot(fpr,tpr,label=f'AUC={auc_score:.2f}')
plt.plot([0,1],[0,1],linestyle='--',color='red')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
plt.rcParams['figure.figsize'] = [6,4]
sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix — Employee Attrition")
plt.show()
```
